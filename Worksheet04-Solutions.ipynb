## Task 2: Temperature converter (write + test)
1. Write `c_to_f(temp_c)` that converts Celsius to Fahrenheit: $F = \frac{9}{5}C + 32$
2. Write `f_to_c(temp_f)` that converts the other way
3. The asserts below check both functions AND that they invert each other.

````python
def c_to_f(temp_c):
    return 9/5 * temp_c + 32

def f_to_c(temp_f):
    return (temp_f - 32) * 5/9

assert c_to_f(0) == 32
assert c_to_f(100) == 212
assert f_to_c(32) == 0
assert abs(f_to_c(c_to_f(37.5)) - 37.5) < 1e-9
print("All conversion tests passed")
````

## Task 4: Refactor repetitive code into a function
The code below computes the kinetic energy for three objects with copy-pasted lines — and one of the copies contains a bug.

1. Find the bug.
2. Replace the repetition with a single function `kinetic_energy(m, v)` and call it three times.
3. Why is the function version safer? (One sentence — a checkpoint favourite.)

````python
# 1. The bug: ke3 uses v3*2 instead of v3**2 (the third copy).
#    With v3 = 3.0 the buggy line prints 9.0, the correct one 13.5 - running the refactor confirms the fix.
#    Note for TAs: with v3 = 2.0 the bug would be INVISIBLE (2*2 == 2**2) - worth mentioning at checkpoint.
# 2. Refactored version: the formula lives in ONE place.
def kinetic_energy(m, v):
    return 0.5 * m * v**2

m1, v1 = 2.0, 3.0
m2, v2 = 1.5, 4.0
m3, v3 = 3.0, 3.0

ke1 = kinetic_energy(m1, v1)
ke2 = kinetic_energy(m2, v2)
ke3 = kinetic_energy(m3, v3)
print(ke1, ke2, ke3)     # 9.0 12.0 13.5

# 3. Why safer (one sentence): the formula exists only once, so it can only be typed
#    wrong once - and a fix or a test in that one place covers every call.
````

## Task 5: A function that returns two values
Write `min_max(values)` that takes a list of numbers and returns **both** the smallest and largest value (in that order), using a single loop — no `min()`/`max()` built-ins.

```python
lo, hi = min_max([3, 1, 4, 1, 5])   # lo = 1, hi = 5
```

````python
def min_max(values):
    # Start both from the FIRST element (not from 0!): a list of only negative
    # numbers would otherwise report hi = 0, which is not in the list.
    lo = values[0]
    hi = values[0]
    for v in values:
        if v < lo:
            lo = v
        if v > hi:
            hi = v
    return lo, hi      # two values come back as one tuple; the caller unpacks them

lo, hi = min_max([3, 1, 4, 1, 5])
assert (lo, hi) == (1, 5)
lo, hi = min_max([-7])
assert (lo, hi) == (-7, -7)     # one element: it is both the smallest and the largest
print("min_max works!")

# Discussion (TA card Q22): min_max([]) raises an IndexError at values[0] - there is no
# smallest element of nothing. A guard clause (print a message, return None) would be the
# Homework 04 style answer; the worksheet does not require it.
````
