## Task 2: Euler's method for motion
Euler's method is a simple numerical technique for solving ordinary differential equations (ODEs) describing motion or change over time. Over the next few weeks we will use increasingly sophisticated numerical techniques to solve increasingly complex physical systems.

Today we will tackle 1D motion for situations where your first-year mechanics knowledge is more than sufficient to allow you to double-check the results.

Let's start with our fundamental definitions of velocity and acceleration, in differential form:

$$\frac{d\vec{r}}{dt} = \vec{v} \;\;\;\;\mbox{and}\;\;\;\; \frac{d\vec{v}}{dt} = \vec{a}.$$

For an object with constant acceleration and known initial conditions, we can know position and velocity for all times, but things can get much more complicated if the acceleration is not constant with time. A great example of this would be an object orbitting a much larger object in a non-circular path.

Using the definition of a derivative for a small $\Delta t$, we could write the above as:

$$\vec{r}(t+\Delta t) \approx \vec{r}(t) + \Delta t \cdot \vec{v}(t),$$

and

$$\vec{v}(t+\Delta t) \approx \vec{v}(t) + \Delta t \cdot \vec{a}(t).$$

Here, $\approx$ reminds us that ratio of the left side to the right side approaches 1 as $\Delta t \rightarrow 0$. The formulas looks suspiciously similar to our first-year constant-acceleration kinematics equations, but we can use these ones even if the acceleration is not constant.

Using numerical methods, if we know position and velocity at a specific time, we can find them a slightly later time $(t+\Delta t)$ and can use those equations over and over again to find positions and velocities at any time in the future.

The solutions will always be approximations, which become more accurate the smaller we make $\Delta t$. As we move toward Project 2, we will also encounter more sophisticated algorithms for solving ordinary differential equations.

### Your sub-tasks
Below is a chunk of code representing a simple physical situation using Euler's method for motion. Some of the steps in here might seem unnecessary, but once we start adding additional complications---such as acceleration---they will help ensure we do things in the correct order.

````python
# Constants and initial values
t = 0.0 # s
dt = 0.1 # s
t_max = 100.0 # s

y = 0. # m
v = 15.3 # m/s

# Calculate the number of steps
n_steps = int(t_max / dt)

for i in range(n_steps):
    
    # Calculate the kinematic variables at t+dt
    y_next = y + v * dt

    # Update time
    t += dt

    # Update kinematic variables for the next step
    y = y_next
    
    if y >= 100:
        print("Break condition achieved")
        break
````

**Task 2a)** Describe the physical system being modelled by this code.

**Model answer.** An object starting at `y = 0` m and moving with a constant velocity of
`v = 15.3` m/s in the `+y` direction. There is no acceleration in the code, so the velocity
never changes: this is 1D constant-velocity motion, stepped forward in time with Euler's
method in steps of `dt = 0.1` s. The loop would run for `n_steps = 1000` steps (100 s), but
the `break` stops it as soon as the object passes `y = 100` m, at about `t = 6.6` s.

**Task 2b)** What does the `break` keyword accomplish in this code?

**Model answer.** It ends the loop immediately, before the remaining iterations run. Here it
is used as a *stopping condition based on the physics* (the object has passed 100 m) instead
of a stopping condition based on counting steps. Without it the loop would keep stepping to
`t = 100` s. Note the test happens at the END of the body, so the loop always overshoots
slightly: it stops at the first step where `y >= 100`, i.e. `y = 100.98` m, not exactly 100 m.

**Task 2g)** Copy your code into the cell below and add gravity in the opposite direction to the initial velocity. Describe what you observe in your output regarding this new situation.

**Model answer.** With `a = -9.8` m/s² the velocity now falls by `9.8 * dt` every step. The
object slows, stops rising at about `t = 1.56` s and `y ≈ 11.9` m, then falls back and goes
negative — the code happily models an object passing through the floor, because nothing in it
says the floor exists. The old `break` at `y >= 100` never fires, so you need a new stopping
condition: the cell below uses "the object came back to the ground" (`y < 0`). Task 2h then
asks for the other natural one — stop at the top.

*Common bug:* updating `v` from `v_next` **before** computing `y_next`, or forgetting
`v = v_next` entirely (then gravity has no effect at all).


````python
# Constants and initial values
t = 0.0      # s
dt = 0.1     # s
t_max = 100.0 # s
y = 0.       # m
v = 15.3     # m/s (initial upward velocity)
a = -9.8     # m/s² (gravity acting downward)

# Calculate the number of steps
n_steps = int(t_max / dt)

# Track maximum height for observation
max_height = 0
turning_point_time = 0

for i in range(n_steps):
    
    # Calculate the kinematic variables at t+dt
    y_next = y + v * dt
    v_next = v + a * dt
    
    # Track maximum height
    if y_next > max_height:
        max_height = y_next
        turning_point_time = t + dt
    
    # Update time
    t += dt
    
    # Update kinematic variables for the next step
    y = y_next
    v = v_next
    
    if y >= 100:
        print(f"Break condition achieved at t={t:.1f}s")
        break
    
    # Check if object falls back to ground
    if y < 0 and t > 0.1:
        print(f"Object returned to ground at t={t:.1f}s")
        break

print(f"\nMaximum height reached: {max_height:.2f}m at t={turning_point_time:.1f}s")
print(f"Final position: y={y:.2f}m")
print(f"Final velocity: v={v:.2f}m/s")
````

**Task 2h)** Modify the `break` condition so that it instead causes the loop to end once the object has reached its maximum height.

**Task 2h) — what to look for.** The natural condition is `if v <= 0: break`, tested after
the velocity update. Groups often write `if v == 0` (never true with `dt = 0.1`) or
`if y_next < y` (also correct, and worth praising — it detects the turn from the position).
The break time lands one `dt` late, which is the whole point of Task 2i.


````python
# Task 2h) solution — break when the object reaches maximum height.
# Maximum height is where the velocity changes sign: the first step with v <= 0.
# (Because we step in finite dt, "v == 0" almost never happens exactly — test for
#  the sign change, never for equality with 0.0.)
t = 0.0
dt = 0.1
t_max = 100.0

y = 0.
v = 15.3
a = -9.8   # m/s^2, gravity opposing the initial velocity

n_steps = int(t_max / dt)

for i in range(n_steps):

    y_next = y + v * dt
    v_next = v + a * dt

    t += dt
    y = y_next
    v = v_next

    if v <= 0:                      # <-- the new break condition
        print(f"Maximum height reached: y = {y:.3f} m at t = {t:.2f} s")
        break

# Compare with the constant-acceleration prediction (this is Task 2i)
t_analytic = 15.3 / 9.8
y_analytic = 15.3**2 / (2 * 9.8)
print(f"Kinematics says:            y = {y_analytic:.3f} m at t = {t_analytic:.3f} s")
print(f"Timing error = {t - t_analytic:.3f} s — always less than one dt.")

# Careful (this is the Task 2i punchline): the break time can only land on a grid time,
# so it improves in JUMPS. dt = 0.4, 0.2, 0.1 and 0.05 all report t = 1.60 s here, and only
# dt = 0.01 gets to 1.57 s. The HEIGHT at the break improves smoothly: 15.07, 13.50, 12.72,
# 12.33 m against the exact 11.94 m. Different errors, different convergence — worth saying out loud.
````

**Task 2i)** Using constant-acceleration kinematics equations, incorporate into your code a calculation for the time at which the object should reach the maximum height. Investigate how close this calculated time is to the time of the `break` condition. How much does this difference depend on your initial velocity? How small do you need to make `dt` before you start to feel like the time of the `break` condition gives a reasonably accurate value? (*Note: you may wish to comment out and/or modify some of your print statements to keep the quantity of the output to a manageable level.*)

````python
# Test different timestep values
dt_values = [1.0, 0.5, 0.1, 0.05, 0.01, 0.005, 0.001]

# Initial velocity for all trials
v_initial = 15.3  # m/s

# Analytical calculation (same for all trials)
t_max_analytical = v_initial / 9.8
y_max_analytical = v_initial * v_initial / (2 * 9.8)

print("=" * 60)
print(f"Analytical time to max height: {t_max_analytical:.4f} s")
print(f"Analytical max height: {y_max_analytical:.2f} m")
print("=" * 60)

# Store results for summary table
results = []

# Loop through each dt value
for dt in dt_values:
    print(f"\n--- Testing dt = {dt} s ---")
    
    # Reset initial values for each trial
    t = 0.0
    t_max = 100.0
    y = 0.
    v = v_initial
    g = -9.8
    
    # Calculate the number of steps
    n_steps = int(t_max / dt)
    
    # Track maximum height
    max_height = 0
    max_height_time = 0
    
    for i in range(n_steps):
        # Calculate the kinematic variables at t+dt
        y_next = y + v * dt
        v_next = v + g * dt
        
        # Track maximum height
        if y_next > max_height:
            max_height = y_next
            max_height_time = t + dt
        
        # Update time
        t += dt
        
        # Update kinematic variables for the next step
        y = y_next
        v = v_next
        
        if y >= 100:
            print(f"Break condition achieved at t={t:.1f}s")
            break
        
        # Check if object falls back to ground
        if y < 0 and t > 0.1:
            break
    
    # Calculate and display errors for this dt
    time_error = abs(max_height_time - t_max_analytical)
    percent_error = (time_error / t_max_analytical) * 100
    
    print(f"Numerical max height time: {max_height_time:.4f} s")
    print(f"Time error: {time_error:.4f} s")
    print(f"Percentage error: {percent_error:.2f}%")
    
    # Store results for summary
    results.append([dt, max_height_time, time_error, percent_error])

# Summary table
print("\n\nSUMMARY:")
print("-" * 60)
print(f"{'dt (s)':<10} {'Numerical t_max (s)':<20} {'Error (s)':<12} {'Error %':<10}")
print("-" * 60)

for dt, t_num, error, percent in results:
    print(f"{dt:<10.4f} {t_num:<20.4f} {error:<12.4f} {percent:<10.2f}")
````

## Task 5 (Optional bonus): the Zen of Python — solution notes

**5a)** `import this` prints the Zen of Python by Tim Peters. It has been in the
standard library since 2004 (PEP 20). Lines that match the mistakes we see most:
*"Explicit is better than implicit"*, *"Errors should never pass silently"*,
*"Readability counts"*, *"Simple is better than complex"*.

The caching gotcha is real and worth letting them hit: `import this` prints only the
**first** time in a kernel, because the module body runs once and is then cached in
`sys.modules`. To see it again: restart the kernel, or `import importlib;
importlib.reload(this)`.

**5b)** `this.__file__` gives the path of the module source, e.g.
`/opt/conda/lib/python3.11/this.py` on Jupyter Open (the exact path depends on the
Python install — that is the point of asking for `__file__` rather than a fixed path).

**5c)** The answers:

1. `65` is `ord("A")` and `97` is `ord("a")` — the ASCII code points where the
   uppercase and lowercase alphabets start. `chr()` is the inverse of `ord()`.
2. The double loop builds a **substitution table**: for each of the two alphabet
   starts, and each offset `i` in `range(26)`, it maps letter `i` to letter `i+13`.
   `d` ends up with **52** entries. `d["a"] == "n"`, `d["n"] == "a"`.
3. `% 26` makes the alphabet **wrap around**. Without it, `chr(i+13+c)` runs past
   `z` into punctuation and other code points — `n` through `z` would decode to
   garbage instead of `a` through `m`.
4. `d.get(c, c)` returns the mapped letter **if `c` is a key**, and otherwise
   returns `c` unchanged. The keys are only the 52 letters, so spaces, newlines,
   commas, apostrophes, `*` and `-` pass through untouched. `d[c]` would raise
   `KeyError` on the first space.
5. It is **ROT-13**: a Caesar cipher with a shift of 13. Because 13 is exactly half
   of 26, the transformation is its own inverse — applying it **twice** returns the
   original text. That is why one table serves for both encoding and decoding, and
   it is the whole reason 13 was chosen.

**5d)** One encoder, reusing the module's own idea:

```python
def rot13(text):
    d = {}
    for c in (65, 97):
        for i in range(26):
            d[chr(i + c)] = chr((i + 13) % 26 + c)
    return "".join([d.get(ch, ch) for ch in text])

msg = "Hello, PHYS 210!"
enc = rot13(msg)
print(enc)          # Uryyb, CULF 210!
print(rot13(enc))   # back to Hello, PHYS 210! - same function, applied twice
assert rot13(enc) == msg
```

Python also ships ROT-13 as a built-in codec, which is a nice punchline once they
have built it themselves:

```python
import codecs
print(codecs.encode("Hello, PHYS 210!", "rot13"))   # Uryyb, CULF 210!
```

Teaching note: this bonus is deliberately an exercise in **reading** code rather than
writing it, and every construct in `this.py` is from Preclass 02 and 03 — strings,
a dictionary, nested `for`, `range()`, indexing and `.get()`. It also rewards the
habit we pushed in lecture: answer the questions by running small experiments, not by
guessing.


````python
# Task 5 solution code
import this
print("module file:", this.__file__)

d = {}
for c in (65, 97):
    for i in range(26):
        d[chr(i + c)] = chr((i + 13) % 26 + c)

print("entries in d:", len(d))
print('d["a"] =', d["a"], '   d["n"] =', d["n"])

def rot13(text):
    return "".join([d.get(ch, ch) for ch in text])

msg = "Hello, PHYS 210!"
enc = rot13(msg)
print(msg, "->", enc, "->", rot13(enc))
assert rot13(enc) == msg
print("ROT-13 is its own inverse")
````
