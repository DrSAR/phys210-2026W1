## Task 1: Fun with arrays
Build a function `my_sign(a)` that takes a NumPy array as an argument and then returns an array of the same shape with the following elements:
* a `1` corresponding to a positive value for that array element in `a`,
* a `-1` corresponding to a negative value for that array element in `a`,
* a `0` corresponding to a zero for that array element in `a`.

No loops: use a mask or `np.where`. Test it on `np.array([-4, 4, 2, 0, 3])` and on a 2-D array.
(The `my_sign` in Preclass 06, section 6.6, crashes when you pass it an array. Why? Be ready to explain it to a TA.)


````python
# Your code here

````

````python
## Solution
import numpy as np

# Why the Preclass 6.6 version crashes: `if a > 0:` on an array gives one True/False
# per element, and `if` needs exactly one -> ValueError (truth value ... is ambiguous).
def my_sign(a):
    return np.where(a > 0, 1, np.where(a < 0, -1, 0))

print(my_sign(np.array([-4, 4, 2, 0, 3])))
print(my_sign(np.array([[5, -3, 2], [9, 0, -2]])))
assert np.array_equal(my_sign(np.array([-4, 4, 2, 0, 3])), [-1, 1, 1, 0, 1])
assert np.array_equal(my_sign(np.array([[5, -3, 2], [9, 0, -2]])), [[1, -1, 1], [1, 0, -1]])
print("Task 1 passes")
````

````python
## Solution
import numpy as np

names = ['Kamala', 'America', 'Cassie', 'Kate', 'Riri', 'Miles']
grades = np.array([[93.0,  79.0,  67.5,  67,  72,  45],
                   [93.0,  79.0,  78.0,  85,  91,  74],
                   [100.0, 92.0,  98.0,  83,  82,  93],
                   [87.0,   0.0,  77.0,   0,  65,  87],
                   [40.0,  45.0,  38.0,  42,  48,  30],
                   [72.0,  80.0,  85.0,  78,  81,  90]])

# 2a) per-student means: axis=1 collapses the columns
hw = grades[:, :3].mean(axis=1)
quiz = grades[:, 3:].mean(axis=1)
overall = 0.3 * hw + 0.7 * quiz
for i in range(len(names)):
    print(f"{names[i]} has an average of {hw[i]:.1f}% on homework, "
          f"{quiz[i]:.1f}% on quizzes, and {overall[i]:.1f}% overall.")

# 2b) per-assessment class means: axis=0 collapses the rows
col_mean = grades.mean(axis=0)
labels = ['HW1', 'HW2', 'HW3', 'Q1', 'Q2', 'Q3']
print("class means:", np.round(col_mean, 1))
print("lowest class mean:", labels[np.argmin(col_mean)])

# 2c) mask
passed = overall >= 50
print("passed:", passed.sum(), "of", len(names))
for i in range(len(names)):
    if passed[i]:
        print("  ", names[i])
# one-liner alternative: print(np.array(names)[passed])

# 2d) letters with nested np.where
letter = np.where(overall >= 80, 'A',
         np.where(overall >= 65, 'B',
         np.where(overall >= 50, 'C', 'F')))
for i in range(len(names)):
    print(names[i], letter[i])

assert abs(overall[0] - 66.9) < 0.05
assert passed.sum() == 5
assert list(letter) == ['B', 'A', 'A', 'C', 'F', 'A']
print("Task 2 passes")
````

## Task 3: Hidden treasure — a safe-path map with masks
At a Rivendell reunion, Gandalf becomes concerned about Pippin's pipeweed use and hides it all. Hobbit feet only tolerate a narrow range of *magical heat*, so Gandalf casts a heat spell over his 20 m × 10 m room. With $x$ and $y$ in metres, the heat at $(x, y)$ is

$$\text{MH} = \cos(x) + \sin(y) \quad \text{(in Rings, Ri)}.$$

Hobbit feet are safe only where $-1.0\ \text{Ri} < \text{MH} < 1.0\ \text{Ri}$. Pippin needs a map.

**Task 3a)** Build the room on a 1-metre grid: `x = np.arange(0, 21)`, `y = np.arange(0, 11)`, then `X, Y = np.meshgrid(x, y)` and `MH = np.cos(X) + np.sin(Y)`. Predict the shape of `MH` before you print it — which axis is $x$?

**Task 3d)** Which $x$ values (columns) are safe for **every** $y$? Use `safe.all(axis=0)` and index `x` with the result. Also print, for every column, how many safe cells it has (`safe.sum(axis=0)`).

**Bonus (only if you have already met matplotlib):** `import matplotlib.pyplot as plt; plt.imshow(safe, origin='lower', extent=(0, 20, 0, 10)); plt.colorbar()` — does the picture match your text map? (Plotting is Day 07; this is optional today.)


````python
# Task 3 code here

````

````python
## Solution
import numpy as np

x = np.arange(0, 21)
y = np.arange(0, 11)
X, Y = np.meshgrid(x, y)
MH = np.cos(X) + np.sin(Y)
print("shape:", MH.shape)            # (11, 21): rows follow y, columns follow x

safe = (MH > -1.0) & (MH < 1.0)
print("safe cells:", safe.sum(), "of", safe.size, f"({safe.mean():.0%})")

chars = np.where(safe, '.', '#')
for row in chars[::-1]:              # last row = largest y, printed first
    print(''.join(row))

always_safe = safe.all(axis=0)
print("x values safe for every y:", x[always_safe])
print("safe cells per column:", safe.sum(axis=0))

assert MH.shape == (11, 21)
assert safe.sum() == 150
assert list(x[always_safe]) == [11]
print("Task 3 passes")
````

Which of `row`, `col`, `big`, `flat` are **views** of `g` and which are **copies**? How would you take row 1 out of `g` so that changing it can never change `g`?


````python
# Prediction of g:
#
#
# Run and compare

````

````python
## Solution
import numpy as np
g = np.arange(12).reshape(3, 4)
row = g[1]
col = g[:, 0]
big = g[g > 6]
flat = g.reshape(12)
row[0] = 100
col[2] = 200
big[0] = 300
flat[-1] = 400
print(g)
# row, col, flat are views (they write into g); big is a copy (mask indexing copies).
# Safe extraction: row_copy = g[1].copy()
print("views:", row.base is not None, col.base is not None, flat.base is not None,
      "| mask is a copy:", big.base is None)
assert np.array_equal(g, [[0, 1, 2, 3], [100, 5, 6, 7], [200, 9, 10, 400]])
print("Task 4 passes")
````

Make `my_sign(a)` from Task 1 also work when a plain number is passed in, returning a plain number (not a 0-d array). Hints: `np.asarray()`, and the `.ndim` of what `np.where` returns.


````python
# Your code here

````

````python
## Solution — bonus: accept an array OR a plain number
import numpy as np

def my_sign(a):
    arr = np.asarray(a)                       # a number becomes a 0-d array
    out = np.where(arr > 0, 1, np.where(arr < 0, -1, 0))
    if out.ndim == 0:                         # plain number in -> plain number out
        return int(out)
    return out

print(my_sign(np.array([-4, 0, 2.5])))        # [-1  0  1]
print(my_sign(-3), my_sign(0), my_sign(7.2))  # -1 0 1
assert type(my_sign(-3)) is int and my_sign(0) == 0 and my_sign(7.2) == 1
assert my_sign(np.array([[5, -3], [0, 2]])).tolist() == [[1, -1], [0, 1]]
print("bonus passes")
````
