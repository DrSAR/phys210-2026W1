## Task 2: Area under a curve by counting points
Estimate the integral

$$I = \int_0^2 e^{-x^2}\,dx$$

with the "fraction of points under the curve" method from the reading (Your turn #1) and from today's slides.
1. Draw `N` random points in the box $0 \le x < 2$, $0 \le y < 1$. Why is a box height of 1 enough here?
2. Count the points with `y < np.exp(-x**2)`.
3. Turn that fraction into an estimate of $I$.

Write this as a function `mc_area(N)` that returns the estimate. Check: the exact value is 0.88208 (5 digits). With `N = 100000` your estimate should be within about 0.01 of it.

Then explain in one sentence (as a comment): why does the fraction of points under the curve, times the area of the box, estimate the integral?

````python
# Task 2 — solution
import numpy as np

def mc_area(N):
    x = np.random.uniform(0, 2, N)
    y = np.random.uniform(0, 1, N)
    under = y < np.exp(-x**2)       # boolean array: True for points under the curve
    fraction = np.sum(under) / N
    return fraction * 2.0           # box area = 2 wide x 1 tall

np.random.seed(1)
print(mc_area(100000), "exact 0.88208")

# Height 1 is enough: exp(-x**2) <= 1 for all x, so the box contains the whole curve.
# Why it works: the points are spread uniformly over the box, so the fraction that lands
# under the curve -> (area under the curve) / (area of the box). Times the box area = integral.
````

## Task 3: Ten times the samples — how much smaller is the scatter?
One Monte-Carlo estimate is one random number: run `mc_area` again and you get a different answer. To see how much it scatters, repeat it.
1. Call `mc_area(1000)` 50 times and store the 50 estimates in a list. Print their mean and their standard deviation (`np.std`).
2. Do the same with `N = 10000` and with `N = 100000`.
3. By roughly what factor does the standard deviation shrink each time you use 10× more samples? Compare with $1/\sqrt{N}$.
4. How many samples would you need to make the scatter 10× smaller than with `N = 1000`?

````python
# Task 3 — solution
for N in [1000, 10000, 100000]:
    estimates = []
    for i in range(50):
        estimates.append(mc_area(N))
    print(N, np.mean(estimates), np.std(estimates))

# Each 10x in N shrinks the std by about sqrt(10) = 3.2 (expected about 0.031, 0.0099, 0.0031).
# With 50 estimates the std itself scatters by about 10%, so "about 3" is the answer, not exactly 3.16.
# Scatter ~ 1/sqrt(N): 10x smaller scatter than N = 1000 needs 100x more samples, N = 100000.
````

## Task 4 (optional): Triangular Plinko
Plinko is a game from the TV show *The Price is Right*. A puck drops through a board of pegs. At each peg it bounces left or right with equal probability, and at the bottom it lands in one of the buckets. On a triangular board the puck can never hit a wall.

![Triangular Plinko](https://i.ibb.co/28pGMgb/triangular-plinko.png)

*(Image captured from https://phet.colorado.edu/sims/html/plinko-probability/latest/plinko-probability_en.html)*

Simulate a triangular board with 8 rows of pegs and 9 buckets.
1. One puck makes 8 random left/right steps; its bucket number (0–8) is the number of steps to the right. Simulate `N = 10000` pucks. One way: `np.random.randint(0, 2, (N, 8))` gives one row of 0s (left) and 1s (right) per puck — sum each row.
2. Make a bar chart (`plt.bar`) or histogram (`plt.hist`) of how many pucks land in each bucket. The plotting examples at the end of this worksheet show both.
3. Compare with the expected counts from the binomial distribution: bucket $k$ should get about $N \binom{8}{k} / 2^8$ pucks, where `math.comb(8, k)` gives $\binom{8}{k}$. Add the expected counts to your plot as points.

````python
# Task 4 — solution
import numpy as np
import matplotlib.pyplot as plt
from math import comb

N = 10000
steps = np.random.randint(0, 2, (N, 8))   # one row per puck: 0 = left, 1 = right, 8 rows of pegs
buckets = np.sum(steps, axis=1)           # bucket 0..8 = number of steps to the right

counts = []
expected = []
for k in range(9):
    counts.append(np.sum(buckets == k))
    expected.append(N * comb(8, k) / 2**8)

plt.bar(range(9), counts, label='simulation')
plt.plot(range(9), expected, 'ro', label='expected (binomial)')
plt.xlabel('bucket')
plt.ylabel('number of pucks')
plt.title(f'Triangular Plinko, {N} pucks')
plt.legend()
plt.show()
````

## Bonus (optional): Rectangular Plinko with bumpers
In the TV-show version the board is rectangular, with 9 possible start positions and 9 buckets. Bumpers on the walls push the puck back: next to a wall the puck has only one way to go instead of two.

![Plinko](https://i.ibb.co/sqqpg9G/plinko.png)

*(Image captured from https://games.washingtonpost.com/games/price-is-right-plinko)*

The 9 start positions are the gaps in the top row (between neighbouring pegs, and between the outer pegs and the walls). After this first row (no randomness yet) the puck passes 12 more rows of pegs where it goes left or right at random. The bucket prizes, left to right, are \$100, \$500, \$1000, \$0, \$10000, \$0, \$1000, \$500, \$100.

Build a Monte-Carlo simulation where you choose the start position and then run $N$ pucks:
1. Make a bar chart or histogram of how many pucks land in each bucket, for at least three start positions (or one heat map: start position vs final bucket).
2. Calculate the expected (average) winnings for each start position. Show them in a plot or with print statements. Where should you drop the puck?

````python
# Bonus — solution
"""
Positions: even numbers 0-16 are the 9 gaps of a row with 9 gaps (start row and bucket row),
odd numbers 1-15 are the 8 gaps of the rows in between. Each peg moves the puck by -1 or +1.
At the walls (0 and 16) the bumper allows only one direction.
"""
import numpy as np
import matplotlib.pyplot as plt

final_bins    = [  0,   2,    4, 6,     8, 10,   12,  14,  16]
bucket_values = [100, 500, 1000, 0, 10000,  0, 1000, 500, 100]
starting_bins = [0, 2, 4, 6, 8, 10, 12, 14, 16]
n_pucks = 1000
num_rows = 12          # must be even, so the puck ends on an even position

def simulate_single_puck(start_bin, num_rows=12):
    position = start_bin
    for row in range(num_rows):
        if position == 0:
            position = 1                  # bumper: only to the right
        elif position == 16:
            position = 15                 # bumper: only to the left
        else:
            position += np.random.choice([-1, 1])
    return position

def run_simulation(start_bin, n_pucks=1000, num_rows=12):
    results = np.zeros(n_pucks, dtype=int)
    for i in range(n_pucks):
        results[i] = simulate_single_puck(start_bin, num_rows)
    return results

def average_winnings(results):
    total = 0
    for final_bin in results:
        total += bucket_values[final_bin // 2]
    return total / len(results)

heatmap_data = np.zeros((len(starting_bins), len(final_bins)))
winnings = np.zeros(len(starting_bins))
for i, start_bin in enumerate(starting_bins):
    results = run_simulation(start_bin, n_pucks, num_rows)
    winnings[i] = average_winnings(results)
    for j, final_bin in enumerate(final_bins):
        heatmap_data[i, j] = np.sum(results == final_bin)

plt.imshow(heatmap_data, cmap='hot', aspect='auto', origin='lower')
plt.xticks(range(len(final_bins)), final_bins)
plt.yticks(range(len(starting_bins)), starting_bins)
plt.xlabel('final position')
plt.ylabel('start position')
plt.title('Counts: start position -> final position')
plt.colorbar(label='counts')
plt.show()

plt.bar(starting_bins, winnings)
plt.xticks(starting_bins)
plt.xlabel('start position')
plt.ylabel('average winnings ($)')
plt.title('Average winnings by start position')
plt.show()
for start_bin, w in zip(starting_bins, winnings):
    print(f"start {start_bin:2d}: ${w:7.0f}")
# The middle start (8) pays best; the outer starts pay least.
````

## Plotting examples: bar chart and histogram

````python
# Plotting examples: bar chart and histogram
import numpy as np
import matplotlib.pyplot as plt

# 10,000 random integers from 0 to 4, inclusive
randos = np.random.randint(0, 5, size=10000)

### Plot 1: bar chart

# How many times did each value 0-4 occur?
values, counts = np.unique(randos, return_counts=True)

# values: the label of each bar; counts: the height of each bar
plt.bar(values, counts)
plt.show()

### Plot 2: bar chart with string labels

names = ['Zero', 'One', 'Two', 'Three', 'Four']
plt.bar(names, counts)
plt.show()

### Plot 3: histogram with the bins given

# A histogram bin is the interval between two edges, [left, right):
# it includes the left edge and excludes the right edge
# (only the right-most bin also includes its right edge).
# So for the values 0-4 we need the edges 0, 1, ..., 5: bins=range(6)

plt.hist(randos, bins=range(6), rwidth=0.8)
plt.show()
````
