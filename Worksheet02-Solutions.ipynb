**Steps: (use python code to do all of the steps)**

First letter:
1. Get the number of times ‘o’ appears in the string in between positions 2 and 8 (inclusive) when counting from 1.
    * When indexing a string, remember that indices start counting at 0, so the first letter 'd' is index 0 but is position 1. Keep the difference clear between *position* (how most of us count, starting at 1) and *index* (starting at 0 for the left-most element).
2. This is the _index_ of string A which is the first letter

Second letter:

2. The answer from 1 is the _negative index_ of the second letter. This means that if you calculated `5` in step 1, you are looking for the letter found at `A[-5]`.

Third letter:
1. Create a string that is made up of _every second letter_ from index 14 to index 34 (inclusive). You may wish to use an internet search or Generative AI to figure out how to choose every second letter when slicing. We will learn more about this soon when we are learning about lists.
2. Find the number of times the letter `i` is in this substring and raise this number to the power of 2. This is the _index_ of the third letter

````python
# Code for your answer here

A = 'dsjsoofdjsnwraebseiyiuhjijenrheegrbeqiloieskjbdkb'

# First letter
count_o = A[1:8].count('o')   # positions 2-8 (1-based) are indices 1-7
first_letter = A[count_o].upper()

# Second letter
string_length = len(A)
second_letter_index = (string_length // 7) + 3
second_letter = A[-second_letter_index]

# Third letter
substring = A[14:35:2]
count_i = substring.count('i')
third_letter_index = count_i ** 2
third_letter = A[third_letter_index]

# Fourth letter
count_e = A.count('e')
count_c = A.count('c')
fourth_letter_index = -(count_e + count_c)
fourth_letter = A[fourth_letter_index]

print(first_letter, second_letter, third_letter, fourth_letter)
````

### Bonus A part 1: complications with leap years?

Task 3 started from an age in **hours**. Now if you know the **date** someone was born,
you know exactly which years elapsed, so the leap days look after themselves and you never count
days at all.

Both dates below are ordinary strings in `YYYY-MM-DD` form. Using only slicing, `int()`,
comparisons and `if`/`else` - no calendar library:

* pull the year, month and day out of each string (`int(birth[0:4])` gives the birth year);
* decide whether this year's birthday has **already happened** on `today`. A birthday that falls
  exactly on `today` counts as having happened;
* set `age` to the number of whole years lived.

Then change `birth` to `"2005-09-14"`, `"2005-09-13"`, `"2006-01-31"` and `"1975-04-27"` and
re-run both cells. The self-check knows the right answer for all five dates.

Notice what you never needed: a day count, a 365, or a leap-year rule.


````python
today = "2026-09-14"   # the day of this class
birth = "2005-09-15"   # then try "2005-09-14", "2005-09-13", "2006-01-31", "1975-04-27"

birth_year = int(birth[0:4])
birth_month = int(birth[5:7])
birth_day = int(birth[8:10])
today_year = int(today[0:4])
today_month = int(today[5:7])
today_day = int(today[8:10])

# The birthday has happened if its month is already past, or if it is this month
# and its day is today or earlier. The boundary case (birthday today) counts as
# "already had it", which is why the test is <= and not <.
if birth_month < today_month:
    had_birthday = True
elif birth_month == today_month and birth_day <= today_day:
    had_birthday = True
else:
    had_birthday = False

if had_birthday:
    age = today_year - birth_year
else:
    age = today_year - birth_year - 1

print(f"Someone born on {birth} is {age} years old on {today}.")
````

### Bonus A part 2: now in hours - with the leap days put back

Part 1 needed no leap-year rule at all. Ask for the age in **hours** and you need one, because
now you have to know how many February 29ths actually went by - and thanks to the birth date,
you do: the years have names, and a named year either contains a February 29 or it does not.

Count them with arithmetic - no calendar library, and no loop, which you have not met yet.
The rule the calendar actually uses is:

> a year is a leap year if it divides by **4** - except a century year, which is not - except
> a century year that also divides by **400**, which is.

So 2024 was a leap year, 1900 was not, and 2000 was.

HINT 1: You cannot walk through the years one at a time to apply that. Count them instead: floor
division (`//`) tells you how many whole multiples of a number fit into another number, and
each "except" in the rule is just another set of multiples to take away or put back. Once you
can say how many leap years there have been up to the end of any year, the leap days inside a
stretch of years are the difference between two of those counts.

HINT 2: Work out the age in hours **on the most recent birthday** - exactly `age` whole years after the
birth date, in the year `last_year = birth_year + age`. Each year of that life runs from one
birthday to the next, so which years to count depends on where the birthday sits relative to
March 1:

* birthday in **March to December**: the February 29 inside each year of life belongs to the
  *later* year, so count the leap years in `birth_year + 1` ... `last_year`;
* birthday in **January or February**: it belongs to the *earlier* year, so both ends of the
  range shift back by one: count `birth_year` ... `last_year - 1`.

Then `age_hours` is `(365*age + leap_days) * 24`, and the hours Task 3's leap-free calendar
would have thrown away is `leap_days * 24`.

SMALL PRINT: A day is taken as exactly 24 hours - no daylight saving, no time zones. 
And someone born on February 29 has no birthday in most years, so "their most recent birthday" is not even defined; 
I don't know what to do about that so stick to the five dates above.

````python
# Bonus A, part 2 - carries on from the cell above, and uses age, birth_year and birth_month

last_year = birth_year + age

# y//4 - y//100 + y//400 counts the leap years from year 1 up to and including y,
# so a difference of two of these counts is the leap days inside a range of years.
# Each year of life runs birthday to birthday, so the range depends on whether the
# birthday falls before or after the February 29 it is paired with.
if birth_month >= 3:
    # birthday in Mar-Dec: the Feb 29 of each year of life belongs to the LATER year
    leaps_to_end = last_year//4 - last_year//100 + last_year//400
    leaps_to_start = birth_year//4 - birth_year//100 + birth_year//400
else:
    # birthday in Jan-Feb: the Feb 29 belongs to the EARLIER year, so shift both ends back one
    leaps_to_end = (last_year - 1)//4 - (last_year - 1)//100 + (last_year - 1)//400
    leaps_to_start = (birth_year - 1)//4 - (birth_year - 1)//100 + (birth_year - 1)//400

leap_days = leaps_to_end - leaps_to_start
age_hours = (365*age + leap_days) * 24
missed_hours = leap_days * 24

print(f"Someone born on {birth} had lived {age_hours} hours by their birthday in {last_year}.")
print(f"A leap-free calendar misses {missed_hours} of them.")

# MAINTAINER NOTE. The five digests in the self-check below are sha256 of f"{birth}-{age_hours}"
# and they depend on `today` (through `age`). If the class date at the top of this bonus changes,
# regenerate them: for each of the five birth dates, run this cell and print
# hashlib.sha256(f"{birth}-{age_hours}".encode()).hexdigest().
````

### Bonus B: make the columns line up

Today's first polling question was a field-width question. Here you use field widths for the job they were invented for: a table whose columns line up by themselves, with no spaces typed by hand.

Build three row strings using the **same** format specification for every row:

* the name is left-aligned in a 12-character column: `{name:<12}`
* the number is right-aligned in a 10-character column with 4 decimal places: `{number:>10.4f}`

Every finished row must be exactly 22 characters long - check with `len()`.

Row C is the trap. Its name is 25 characters, and a field width is a *minimum*, not a maximum: python will pad a short name but it will never trim a long one, so row C grows and the column breaks. Trim it yourself first: if the name is longer than 12 characters, use the first 10 characters followed by `..`; otherwise use the name unchanged. That gives `electron r..`, which is exactly 12.

````python
name_a = "tau"
num_a = 6.283185307179586
name_b = "gravity"
num_b = 9.81
name_c = "electron rest mass energy"
num_c = 0.510998950

row_a = f"{name_a:<12}{num_a:>10.4f}"
row_b = f"{name_b:<12}{num_b:>10.4f}"

# A field width is a MINIMUM. It pads a short name but never trims a long one,
# so name_c has to be shortened before it is formatted.
if len(name_c) > 12:
    short_c = name_c[:10] + ".."
else:
    short_c = name_c

row_c = f"{short_c:<12}{num_c:>10.4f}"

print(row_a)
print(row_b)
print(row_c)
````

````python
# Self-check for Bonus B - the spaces matter, so the test is on the exact strings
if row_a is None or row_b is None or row_c is None:
    print("Fill in row_a, row_b and row_c above, then re-run this cell.")
else:
    assert row_a == "tau             6.2832", f"row_a is {row_a!r}"
    assert row_b == "gravity         9.8100", f"row_b is {row_b!r}"
    assert len(row_c) == 22, \
        f"row_c is {len(row_c)} characters, not 22 - the long name is still pushing the number across: {row_c!r}"
    assert row_c == "electron r..    0.5110", f"row_c is {row_c!r}"
    print("All three rows are 22 characters wide, so the decimal points line up.")
````
