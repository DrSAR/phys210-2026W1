## Task 3: while — find the threshold
A capacitor discharges: each millisecond its voltage becomes 95% of the previous value, starting at 5.0 V. Use a `while` loop to find after how many whole milliseconds the voltage first drops below 0.5 V. Check: it's between 40 and 50.

````python
v = 5.0
ms = 0
while v >= 0.5:
    v = v * 0.95
    ms += 1
print(ms, "ms, final voltage", v)   # 45 ms
````

## Task 4: dictionary accumulator — count letters
Count how often each letter occurs in `"computational physics"` (skip the space). Build a dictionary `freq` with ONE loop over the string: for each letter ask *is this key in the dictionary yet?* — if not, create it with 1; if yes, add 1. Then find the largest count and every letter that has it. Is there a single winner?

````python
text = "computational physics"
freq = {}
for ch in text:
    if ch == " ":
        continue
    if ch in freq:
        freq[ch] = freq[ch] + 1
    else:
        freq[ch] = 1
print(freq)

most = max(freq.values())     # the largest count
winners = []
for ch, n in freq.items():
    if n == most:
        winners.append(ch)
print(most, winners)   # 2 ['c', 'o', 'p', 't', 'a', 'i', 's']: a seven-way tie, no single winner
````

## Bonus: while + random draws — loop until it happens
Write a loop that repeatedly picks a random integer 1–100 (`np.random.randint(1, 101)`) until it gets one divisible by 7, counting the attempts. Run it a few times — do the counts vary the way you'd expect? (Teaser for Monte-Carlo on Day 09.)

````python
import numpy as np
attempts = 0
while True:
    attempts += 1
    if np.random.randint(1, 101) % 7 == 0:
        break
print(attempts)   # 14 multiples of 7 in 1..100 -> p = 0.14 per draw -> about 7 attempts on average, but anything from 1 to 20+ is normal
````
