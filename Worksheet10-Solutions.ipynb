```python
# OPTIONAL helper: sliders to find good initial guesses. Run me, then move the sliders.
%matplotlib widget
import ipywidgets as widgets

def guess_model(t, a, b, f):    # the same formula as the model in Task 1
    return a * np.exp(np.sin(2 * np.pi * f * t)) + b

t_smooth = np.linspace(x.min(), x.max(), 400)
fig, ax = plt.subplots(num='sparkle-guess', clear=True, figsize=(7, 4))
ax.errorbar(x, y, yerr=dy, fmt='o', ms=4, label='data')
(guess_line,) = ax.plot(t_smooth, guess_model(t_smooth, 1, 0, 0.5), 'r-', label='model with your guesses')
ax.set_xlabel('Time (s)')
ax.set_ylabel('Wand power (arbitrary units)')
ax.legend(loc='upper right')

def update(a=1.0, b=0.0, f=0.5):
    guess_line.set_ydata(guess_model(t_smooth, a, b, f))
    ssr = np.sum((y - guess_model(x, a, b, f))**2)
    ax.set_title(f"p0 = [{a:.2f}, {b:.2f}, {f:.3f}]    sum of squared residuals = {ssr:.1f}")
    fig.canvas.draw_idle()

widgets.interact(update, a=(0.0, 3.0, 0.05), b=(-2.0, 3.0, 0.05), f=(0.05, 1.5, 0.005));
```

```python
%matplotlib inline
# (the line above switches back to normal plots after the optional slider helper)
# Task 1 solution

# Part 1: the model. The independent variable (time) must be the FIRST argument:
# curve_fit calls sparkle(x, a, b, f) with the data x and then varies a, b, f.
def sparkle(t, a, b, f):
    return a * np.exp(np.sin(2 * np.pi * f * t)) + b

plt.errorbar(x, y, yerr=dy, fmt='o', ms=4)
plt.title("Wand data")
plt.xlabel("Time (s)")
plt.ylabel("Wand power (arbitrary units)")
plt.show()

# Part 2: initial guesses, read off the plot.
# - e^sin(...) swings between e^-1 = 0.37 and e^1 = 2.72, a range of 2.35.
#   Most data swing between about 0.75 and 3.6, a range of about 2.9 -> a = 2.9/2.35 = 1.2
# - lowest value a*0.37 + b = 0.75 -> b = 0.75 - 0.45 = 0.3
# - peaks near t = 0.4, 2.0, 3.8, 5.4 s -> period about 1.7 s -> f = 1/1.7 s = 0.6 Hz
p0 = [1.2, 0.3, 0.6]
t_smooth = np.linspace(x.min(), x.max(), 400)

plt.errorbar(x, y, yerr=dy, fmt='o', ms=4, label="data")
plt.plot(t_smooth, sparkle(t_smooth, *p0), "r-", label=f"model with p0 = {p0}")
plt.title("Initial guesses")
plt.xlabel("Time (s)")
plt.ylabel("Wand power (arbitrary units)")
plt.legend(loc="upper right")
plt.show()

# Part 3: the fit, with the measurement uncertainties
popt, pcov = optimize.curve_fit(sparkle, x, y, p0=p0, sigma=dy, absolute_sigma=True)
a_fit, b_fit, f_fit = popt
da, db, df = np.sqrt(np.diag(pcov))
print(f"a = {a_fit:.3f} ± {da:.3f}")
print(f"b = {b_fit:.3f} ± {db:.3f}")
print(f"f = {f_fit:.4f} ± {df:.4f} Hz")

# Part 4: data + fit
plt.errorbar(x, y, yerr=dy, fmt='o', ms=4, label="data")
plt.plot(t_smooth, sparkle(t_smooth, *popt), "r-", label="best fit")
plt.title("Gandalf's wand: best fit")
plt.xlabel("Time (s)")
plt.ylabel("Wand power (arbitrary units)")
plt.legend(loc="upper right")
plt.show()

# Part 5: compare with 0.4 Hz
n_sigma = (f_fit - 0.4) / df
print(f"f differs from 0.4 Hz by {f_fit - 0.4:.4f} Hz = {n_sigma:.0f} uncertainties")
# About 200 uncertainties apart: the spell runs at about 0.59 Hz, not 0.4 Hz.
# The wand is misbehaving (or Gandalf's 0.4 Hz is wrong).
```

## Task 2: Residuals — and a bad start
The residuals are the data minus the model: `y - sparkle(x, *popt)`, one number per data point.

1. Compute the residuals of your Task 1 fit. Plot them against time, with the error bars `dy` and a horizontal line at zero (`plt.axhline(0)`). Put this plot under your Task 1 plot (for example with `fig, (ax1, ax2) = plt.subplots(2, 1, sharex=True)`), or make a new figure.
2. What *should* the residuals look like when the model describes the data well? Do yours? Answer as a comment.
3. Gandalf insists on 0.4 Hz. Fit again, now starting from his value: `p0 = [1, 0, 0.4]`. Print $f$ ± its uncertainty. Plot this fit over the data and its residuals, the same way as before.
4. Both fits finish without an error message, and both report a small uncertainty. Which fit is right, and how do the residuals tell you? Why can a bad initial guess `p0` ruin a fit? Answer as comments.

```python
# Task 2 solution

# Part 1: residuals of the good fit, plotted under the fit
res = y - sparkle(x, *popt)

fig, (ax1, ax2) = plt.subplots(2, 1, sharex=True, figsize=(7, 6), gridspec_kw={'height_ratios': [2, 1]})
ax1.errorbar(x, y, yerr=dy, fmt='o', ms=4, label="data")
ax1.plot(t_smooth, sparkle(t_smooth, *popt), "r-", label=f"fit from p0 = {p0}")
ax1.set_ylabel("Wand power (arb. units)")
ax1.legend(loc="upper right")
ax2.errorbar(x, res, yerr=dy, fmt='o', ms=4)
ax2.axhline(0, color="k")
ax2.set_xlabel("Time (s)")
ax2.set_ylabel("Residual")
plt.show()

# Part 2: for a good fit the residuals scatter randomly around zero, with no pattern
# (no waves, no trend, no curvature), and most error bars reach the zero line
# (about 2 out of 3 within one error bar). Ours do.

# Part 3: start from Gandalf's 0.4 Hz
p0_bad = [1, 0, 0.4]
popt_bad, pcov_bad = optimize.curve_fit(sparkle, x, y, p0=p0_bad, sigma=dy, absolute_sigma=True)
f_bad, df_bad = popt_bad[2], np.sqrt(np.diag(pcov_bad))[2]
print(f"start p0 = {p0_bad}:  f = {f_bad:.4f} ± {df_bad:.4f} Hz")
res_bad = y - sparkle(x, *popt_bad)

fig, (ax1, ax2) = plt.subplots(2, 1, sharex=True, figsize=(7, 6), gridspec_kw={'height_ratios': [2, 1]})
ax1.errorbar(x, y, yerr=dy, fmt='o', ms=4, label="data")
ax1.plot(t_smooth, sparkle(t_smooth, *popt_bad), "m-", label=f"fit from p0 = {p0_bad}")
ax1.set_ylabel("Wand power (arb. units)")
ax1.legend(loc="upper right")
ax2.errorbar(x, res_bad, yerr=dy, fmt='o', ms=4, color="m")
ax2.axhline(0, color="k")
ax2.set_xlabel("Time (s)")
ax2.set_ylabel("Residual")
plt.show()

# Part 4: the first fit is right. Its residuals scatter randomly around zero
# (3 of 4 points lie within one error bar of zero).
# The fit from 0.4 Hz stays near 0.39 Hz. Its residuals show structure: long runs
# above and then below zero that drift with time, because the fitted wave slowly
# gets out of step with the data. Most of them are far bigger than their error bars.
# The weighted sum of squared residuals tells the same story:
print(f"sum of (res/dy)^2: good fit {np.sum((res / dy)**2):.0f}, fit from 0.4 Hz {np.sum((res_bad / dy)**2):.0f}, for {len(x)} points")
# Why can a bad p0 ruin a fit? curve_fit starts at p0 and walks downhill in the
# sum of squared residuals. It stops in the nearest valley (a local minimum)
# and cannot tell that a deeper valley exists elsewhere. The small uncertainty
# only describes the shape of THAT valley, so it does not warn you; the residuals do.
```

## Task 3 (optional): Simulate and fit radiation-shielding data
This task is based on a PHYS 119 experiment: you measure the attenuation coefficient $\mu$ of a shielding material for a radioactive source. A simple first-order model for the count rate in the detector is

$$R(x) = R_0\, e^{-\mu x} + R_b,$$

where

* $x$ is the thickness of the shielding between the source and the detector,
* $R_0$ is the count rate from the source without shielding (source at a fixed distance from the detector),
* $R_b$ is the background count rate from the environment; the shielding does not change it, and
* $R(x)$ is the count rate the detector measures: source plus background.

In Task 3.1 you run the experiment (as a simulation). In Task 3.2 you analyse your data.

### Task 3.1: Simulate the measurements
1. Use the true values $R_0 = 20$ counts/s, $R_b = 0.18$ counts/s and $\mu = 0.5\ \mathrm{mm}^{-1}$.
2. Radioactive counts follow a Poisson distribution. If you measure for a time $t$ at thickness $x$, the expected number of counts is $\lambda = R(x)\, t$, and `np.random.poisson(lam)` gives you one simulated measurement. Example: a one-hour (3600 s) measurement of the background alone has $\lambda = 0.18 \times 3600 = 648$ expected counts. For $\lambda$ of about 20 or more, the Poisson distribution looks like a Gaussian with standard deviation $\sqrt{\lambda}$. So one number, $\lambda$, sets both the mean and the spread, and a measured count $N$ has an uncertainty of about $\sqrt{N}$.
3. You have 30 minutes of measuring time in total, shared over all your thicknesses: choose how many thicknesses, and how long to measure at each. Include a few thicknesses where the shielding blocks almost all of the source, so that the background dominates. These points pin down $R_b$, and that gives a better $\mu$.
4. Turn your counts into rates ($N/t$, with uncertainty $\sqrt{N}/t$) and plot the rate against the thickness, with error bars.

```python
# Task 3.1 solution
R0_true, Rb_true, mu_true = 20, 0.18, 0.5      # counts/s, counts/s, 1/mm

def shield(x, R0, mu, Rb):
    return R0 * np.exp(-mu * x) + Rb

np.random.seed(210)                            # reproducible simulation
thick = np.linspace(0, 20, 21)                 # 21 thicknesses from 0 to 20 mm
t_each = 30 * 60 / thick.size                  # 30 minutes shared equally: about 86 s each
lam = shield(thick, R0_true, mu_true, Rb_true) * t_each   # expected counts
counts = np.random.poisson(lam)                # one simulated measurement per thickness
rates = counts / t_each
d_rates = np.sqrt(counts) / t_each
# (at large x a count can be small; the sqrt(N) uncertainty is then only rough)

plt.errorbar(thick, rates, yerr=d_rates, fmt='o', ms=4)
plt.yscale('log')
plt.xlabel("Shielding thickness (mm)")
plt.ylabel("Count rate (counts/s)")
plt.title("Simulated count rates")
plt.show()
```

### Task 3.2: Fit — what is your measured $\mu$?
Because you know the true values that went into the simulation, you can check whether the fit and its uncertainties are honest.

1. Fit the model to your simulated rates with `curve_fit`. Use the rate uncertainties as `sigma`, with `absolute_sigma=True`.
2. Report $\mu$ ± its uncertainty. How many uncertainties is it from the true 0.5 mm⁻¹?
3. Plot your data (with error bars) and the fitted model, and the residuals under it.
4. Fit again with a model *without* the background term, $R_0\, e^{-\mu x}$. Where do its residuals show structure, and what does that tell you?

```python
# Task 3.2 solution
popt_s, pcov_s = optimize.curve_fit(shield, thick, rates, p0=[20, 0.5, 0.2],
                                    sigma=d_rates, absolute_sigma=True)
dR0, dmu, dRb = np.sqrt(np.diag(pcov_s))
print(f"R0 = {popt_s[0]:.2f} ± {dR0:.2f} counts/s")
print(f"mu = {popt_s[1]:.4f} ± {dmu:.4f} 1/mm  ({(popt_s[1] - mu_true) / dmu:+.1f} uncertainties from the true 0.5)")
print(f"Rb = {popt_s[2]:.3f} ± {dRb:.3f} counts/s")

# Model without background
def shield_nobg(x, R0, mu):
    return R0 * np.exp(-mu * x)

popt_n, pcov_n = optimize.curve_fit(shield_nobg, thick, rates, p0=[20, 0.5],
                                    sigma=d_rates, absolute_sigma=True)
print(f"without background: mu = {popt_n[1]:.4f} ± {np.sqrt(pcov_n[1, 1]):.4f} 1/mm")

x_smooth = np.linspace(thick.min(), thick.max(), 200)
fig, (ax1, ax2, ax3) = plt.subplots(3, 1, sharex=True, figsize=(7, 8), gridspec_kw={'height_ratios': [2, 1, 1]})
ax1.errorbar(thick, rates, yerr=d_rates, fmt='o', ms=4, label="simulated data")
ax1.plot(x_smooth, shield(x_smooth, *popt_s), "r-", label="fit with background")
ax1.plot(x_smooth, shield_nobg(x_smooth, *popt_n), "g--", label="fit without background")
ax1.set_yscale('log')
ax1.set_ylabel("Rate (counts/s)")
ax1.legend()
ax2.errorbar(thick, rates - shield(thick, *popt_s), yerr=d_rates, fmt='o', ms=4, color="r")
ax2.axhline(0, color="k")
ax2.set_ylabel("Residual\n(with bg)")
ax3.errorbar(thick, rates - shield_nobg(thick, *popt_n), yerr=d_rates, fmt='o', ms=4, color="g")
ax3.axhline(0, color="k")
ax3.set_ylabel("Residual\n(no bg)")
ax3.set_xlabel("Shielding thickness (mm)")
plt.show()
# With background: residuals scatter around zero, and mu agrees with 0.5 within its uncertainty
# (for most seeds within 1-2 uncertainties).
# Without background: at large thickness (almost) all residuals sit ABOVE zero (the model goes to 0,
# the data level off at Rb) -> structure: the model misses the background.
```
